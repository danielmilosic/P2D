```python
### CONCAT TWO VIDEO + add image

import subprocess

first_start = 0.0
first_end = 83.75
second_start = 45.25

image_start = 10.5
image_end = 15.75

filter_complex = (
    f"[0:v]trim=start={first_start}:end={first_end},"
    f"setpts=PTS-STARTPTS[v0];"

    f"[1:v]trim=start={second_start},"
    f"setpts=PTS-STARTPTS[v1];"

    f"[v0][v1]concat=n=2:v=1:a=0[base];"

    f"[base][2:v]overlay=10:10:"
    f"enable='between(t,{image_start},{image_end})'[v]"
)

cmd = [
    "ffmpeg",
    "-i", "first.mp4",
    "-i", "second.mp4",
    "-loop", "1",
    "-i", "overlay.png",
    "-filter_complex", filter_complex,
    "-map", "[v]",
    "-y",
    "output.mp4",
]

subprocess.run(cmd, check=True)
```
