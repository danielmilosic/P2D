```python
import subprocess
from pathlib import Path


def process_audio(input_file: Path, output_file: Path):
    """
    Slows audio and adds quiet white noise at the beginning.
    """

    ffmpeg_cmd = [
        "ffmpeg",
        "-y",
        "-i", str(input_file),

        "-filter_complex",
        (
            f"[0:a]atempo=0.6,aresample=44100,pan=stereo|c0=c0|c1=c1[slow];"
            f"anoisesrc=color=white:duration=0.5:sample_rate=44100,"
            f"volume=-70dB,pan=stereo|c0=c0|c1=c0[noise];"
            "[noise][slow]concat=n=2:v=0:a=1[out]"
        ),

        "-map", "[out]",
        "-c:a", "libmp3lame",
        "-b:a", "192k",
        str(output_file.with_suffix(".mp3"))
    ]

    subprocess.run(ffmpeg_cmd, check=True)

# CONFIGURATION
INPUT_DIR = Path("input_files")
OUTPUT_DIR = Path("output_files")

SPEED_FACTOR = 0.7          # 60% speed
NOISE_DURATION = 1.5        # seconds
NOISE_VOLUME_DB = -60       # quiet white noise

SUPPORTED_EXTENSIONS = {".mp3", ".wav", ".m4a", ".aac", ".flac"}

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)



for file in INPUT_DIR.iterdir():
    if file.suffix.lower() in SUPPORTED_EXTENSIONS:
        output_file = OUTPUT_DIR / file.name
        print(f"Processing: {file.name}")
        process_audio(file, output_file)

print("Processing complete.")


```
