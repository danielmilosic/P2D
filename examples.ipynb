# P2D Examples

### 1. Quickly loading and plotting in-situ data

```python
from P2D import dataloader
from P2D import P2D_plots
import matplotlib.pyplot as plt

time = ['2025-06-01', '2025-7-1']
spacecraft = 'OMNI'

data = dataloader.load(spacecraft=spacecraft, timerange=time)
P2D_plots.plot_timeseries(data)
```

### 2. Model run with multiple Spacecraft + Plot

```python
from P2D import P2D

timerange=['2022-09-25', '2022-10-30']

model = P2D.P2D(input_data=['OMNI', 'STEREO-A', 'PSP', 'SolO']
                , timerange=timerange
                , propagation_time='4d'
                , model_type='inelastic'
                , degree_resolution=0.5
                , COR = 0
                , smooth = '1d')

model.plot(rlim=3, variable_to_plot='V')
```

### 3. Plot the model results as a timeseries (In Carrington Coordinates)

```python
model.plot_timeseries(variable_to_plot='V')
```

### 4. Cutout the timeseries at a specific spacecraft and plot

```python
model.observe_from(observer='OMNI')
model.plot_timeseries(model.list_of_timeseries_cutouts[0])
```

### 5. Plot against time

```python
model.plot_timeseries(model.list_of_timeseries_cutouts[0], plot_against_time=True, variable_to_plot='V')
```

### 6. Make a movie of the model results 

```python
model.movie(cadence='6h'
            , rlim=3
            , time_window='5d'
            , observers=['OMNI', 'STEREO-A', 'SolO', 'PSP'])
```
