```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates
from scipy.signal import find_peaks
import pandas as pd
from P2D import dataloader

#P2D
# time_series_0 = pd.read_parquet('combined_model_superlong_2007_2008_1d_smooth.parquet')
# time_series_1 = pd.read_parquet('combined_model_superlong_2008_2015_1d_smooth.parquet')
# time_series_2 = pd.read_parquet('single_model_superlong.parquet')
# time_series_3 = pd.read_parquet('combined_model_superlong_2023_2026_1d_smooth.parquet')

# time_series_1 = time_series_1.loc[time_series_0.index[-1]:]
# time_series_2 = time_series_2.loc[time_series_1.index[-1]:]
# time_series_3 = time_series_3.loc[time_series_2.index[-1]:]
# time_series = pd.concat([time_series_0, time_series_1, time_series_2, time_series_3]).drop_duplicates(keep='first')

# #BETTER PEAKS
# time_series_3 = pd.read_parquet('P2D_2023_2026_2h_smooth.parquet')
# time_series_2 = pd.read_parquet('P2D_2014_2023_2h_smooth.parquet')


# time_series_2 = pd.read_parquet('single_model_superlong.parquet')
# time_series_3 = pd.read_parquet('combined_model_superlong_2023_2026_1d_smooth.parquet')

# time_series_3 = time_series_3.loc[time_series_2.index[-1]:]
# time_series = pd.concat([time_series_2, time_series_3]).drop_duplicates(keep='first')

#time_series = time_series.rolling('1d').mean()

#P27
time_series = pd.read_parquet('27_27d_persistence_1h_smooth_whole_range.parquet')

#ESWF
# time_series= pd.read_csv('forecast_archive1.txt')
# time_series['DATE'] = pd.to_datetime(time_series['DATE'])
# time_series['N'] = time_series[' V']
# time_series.set_index('DATE', inplace=True)
# for i in range(2,8):
#     print(i)
#     time_series_addon = pd.read_csv('forecast_archive'+str(i)+'.txt')
#     time_series_addon['DATE'] = pd.to_datetime(time_series_addon['DATE'])
#     time_series_addon.set_index('DATE', inplace=True)
#     time_series_addon['N'] = time_series_addon[' V']
#     time_series_addon = time_series_addon.resample('3H').mean()
#     time_series_addon = time_series_addon.loc[time_series.index[-1]:]
    
#     time_series = pd.concat([time_series, time_series_addon])

#HUXt
# time_series = pd.read_csv('insitu_huxt_forecasts_4day_leadtime.csv')
# time_series['time'] = pd.to_datetime(time_series['time'])
# time_series.set_index('time', inplace=True)
# time_series['N'] = time_series['vsw']
# time_series['Observer_SC'] = 'OMNI'


#timeframe = ['2008-10-15', '2024-1-1']
timeframe = ['2012-06-1', '2024-1-1']
#timeframe = ['2014-10-15', '2024-1-1']
#timeframe = None

if timeframe is not None:
    time_series = time_series[timeframe[0]:timeframe[1]]

start_time = time_series.index.min().strftime('%Y-%m-%d-%H:%M')
end_time = time_series.index.max().strftime('%Y-%m-%d-%H:%M')

# Load spacecraft data
spacecraft_data = dataloader.load(spacecraft='OMNI', timerange=[start_time, end_time], cadence='1h')
spacecraft_data = spacecraft_data[~spacecraft_data.index.duplicated(keep='first')]



hss = pd.read_csv('OMNI_SIR_HSS_1995_2023_Koller2025.txt', sep='   ', skiprows=30, names= ['start', 'peak', 'end', 'source', 'comment', 'comment2'])
hss['start'] = pd.to_datetime(hss['start'])
hss['end']   = pd.to_datetime(hss['end'])
hss['peak']   = pd.to_datetime(hss['peak'])

sns.set(
    style="ticks",
    palette="muted",
    rc={
        "axes.labelsize": 10,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
        "font.size": 10,
        "figure.dpi": 250,
        "legend.fontsize": 8
    }
)

spacecraft_data.dropna(subset=('N'), inplace=True)


# --- GLOBAL PEAK DETECTION ---

# Spacecraft peaks
y_sc = spacecraft_data['N'].values
peaks_sc, _ = find_peaks(y_sc, height=(5, 100), distance=12, prominence=8)
starts_sc =  spacecraft_data.index.get_indexer(hss['start'], method='nearest')
times_sc = spacecraft_data.index[peaks_sc]
vals_sc = y_sc[peaks_sc]

# Start times (convert indices back to timestamps)
start_times = spacecraft_data.index[starts_sc]

# Convert to numpy datetime64 for fast ops
times_np = times_sc.values.astype('datetime64[ns]')
starts_np = start_times.values.astype('datetime64[ns]')

# Compute absolute time differences (broadcasting)
diff = np.abs(times_np[:, None] - starts_np[None, :])

# 24 hours threshold
threshold = np.timedelta64(24, 'h')

# Keep peaks that are within 24h of ANY start
mask = (diff <= threshold).any(axis=1)

# Apply mask
peaks_sc = peaks_sc[mask]
times_sc = times_sc[mask]
vals_sc = vals_sc[mask]


# Time series peaks (align first!)
time_series = time_series[['N']].resample('1h').mean()
ts_aligned = time_series.reindex(spacecraft_data.index).interpolate()

y_ts = ts_aligned['N'].values
peaks_ts, _ = find_peaks(y_ts, height=(5, 100), distance=12, prominence=3)
times_ts = ts_aligned.index[peaks_ts]
vals_ts = y_ts[peaks_ts]

# MATCH PEAKS

threshold = pd.Timedelta('4D')
# Build all candidate pairs within threshold
candidates = []

for i, (t_sc, v_sc) in enumerate(zip(times_sc, vals_sc)):
    for j, (t_ts, v_ts) in enumerate(zip(times_ts, vals_ts)):
        dt = abs(t_ts - t_sc)
        if dt <= threshold:
            candidates.append({
                "i_sc": i,
                "i_ts": j,
                "t_sc": t_sc,
                "v_sc": v_sc,
                "t_ts": t_ts,
                "v_ts": v_ts,
                "dt": dt
            })



# Sort by smallest time difference first
candidates = sorted(candidates, key=lambda x: x["dt"])

# Greedy selection (no duplicates)
used_sc = set()
used_ts = set()
matches = []

for c in candidates:
    if c["i_sc"] in used_sc or c["i_ts"] in used_ts:
        continue

    matches.append(c)
    used_sc.add(c["i_sc"])
    used_ts.add(c["i_ts"])


N=8
M=4

total_chunks = N * M
indices = np.array_split(np.arange(len(spacecraft_data)), total_chunks)

for fig_idx in range(M):
    fig, axes = plt.subplots(nrows=N, ncols=1, figsize=(10, 1.5*N), sharey=False, sharex=False)

    # Get chunks for this figure
    chunk_group = indices[fig_idx * N : (fig_idx + 1) * N]

    # # --- Create N stacked subplots ---
    # fig, axes = plt.subplots(nrows=N, ncols=1, figsize=(15, 40), sharey=False, sharex=False)

    # # Split indices into N chunks
    # indices = np.array_split(np.arange(len(spacecraft_data)), N)

    for ax, idx in zip(axes, chunk_group):
        # Slice data
        sc_chunk = spacecraft_data.iloc[idx]
        ts_chunk = time_series.reindex(sc_chunk.index)

        # Plot lines
        sns.scatterplot(x=sc_chunk.index, y=sc_chunk['N'], ax=ax, linewidth=0, s=5, label='OMNI')
        sns.scatterplot(x=ts_chunk.index, y=ts_chunk['N'], ax=ax, linewidth=0, s=5, label='P2D')

        # Shade HSS intervals overlapping this chunk
        t_min = sc_chunk.index.min()
        t_max = sc_chunk.index.max()

        for _, row in hss.iterrows():
            start = row['start']
            end   = row['end']

            # Only plot if interval overlaps this subplot's time range
            if (end >= t_min) and (start <= t_max):
                ax.axvspan(start, end, color='grey', alpha=0.2, zorder=0)

        for m in matches:
            # Only plot if BOTH points fall inside this chunk
            if (m["t_sc"] in sc_chunk.index) and (m["t_ts"] in ts_chunk.index):
                
                # Plot peak markers
                ax.scatter(m["t_sc"], m["v_sc"], color='green', s=10, zorder=3)
                ax.scatter(m["t_ts"], m["v_ts"], color='red', s=10, zorder=3)

                # --- CONNECT THEM ---

                # Vertical line (time alignment reference)
                ax.plot([m["t_sc"], m["t_sc"]],
                        [m["v_sc"], m["v_ts"]],
                        color='gray', linestyle='--', linewidth=0.8)

                # Horizontal line (time offset visualization)
                ax.plot([m["t_sc"], m["t_ts"]],
                        [m["v_ts"], m["v_ts"]],
                        color='blue', linestyle='--', linewidth=0.8)



        # X-axis formatting
        # ax.xaxis.set_major_locator(mdates.DayLocator((1, 5, 10, 15, 20, 25)))
        ax.xaxis.set_major_locator(mdates.DayLocator((1)))
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
        ax.legend(loc='upper right')
        ax.set_xlabel('Date')
        ax.set_ylim(1,100)
        ax.set_yscale('log')
        ax.set_ylabel(r'N [ $cm^{-3}$]')
        # Only rotate labels on bottom plot (cleaner)
        if ax != axes[-1]:
            ax.set_xlabel('')
        #else:
        #plt.setp(ax.get_xticklabels(), rotation=10, ha='right')

    fig.subplots_adjust(hspace=-0.1)
    plt.tight_layout()
```

```python
import pandas as pd
import seaborn as sns
from P2D import dataloader
from P2D import P2D_plots
import matplotlib.pyplot as plt
from P2D.utils import spacecraft_ID
import numpy as np
from scipy.stats import linregress
sns.set(
    style="ticks",
    palette="muted",
    rc={
        "axes.labelsize": 15,    # Axis label size
        "xtick.labelsize": 15,   # X-axis tick label size
        "ytick.labelsize": 15,   # Y-axis tick label size
        "font.size": 13,         # Overall font size
        "figure.dpi": 100,        # Dots per inch for figure quality
        "legend.fontsize": 15 # Size of legend
    }
)


# OMNI

spacecraft_to_compare = 'OMNI'

res1 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part1.parquet')
res2 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part2.parquet')
res3 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part3.parquet')
res4 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part4.parquet')
res5 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part5.parquet')
res6 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part6.parquet')
res7 = pd.read_parquet('model_cutouts/model_results_omni_2h_smoothed_inelastic_100dwindow_part7.parquet')


res = pd.concat([res1, res2, res3, res4, res5, res6, res7]).drop_duplicates(keep='first')

res = res[res['N']< 50]

#res['Time'] = res.index
#res.set_index('Encounter_time', inplace=True)
time = [res.index[0], res.index[-1]]

omni = dataloader.load(spacecraft=spacecraft_to_compare, timerange=time)
omni = omni[~omni.index.duplicated(keep='first')]
omni = omni.rolling('1d').mean()

remove_cmes=True
if remove_cmes:
    cmes = pd.read_excel('HELIO4CAST_ICMECAT_v23 (1).xlsx')

    spacecraft = ['Wind', 'STEREO-A', 'SolarOrbiter', 'PSP']
    my_spacecraft = ['OMNI', 'STEREO-A', 'SolO', 'PSP']

    for i, sc in enumerate(spacecraft):
        print(sc, ' => ', my_spacecraft[i])
        start_times = pd.to_datetime(cmes.loc[cmes['sc_insitu']==sc]['icme_start_time']).reset_index()['icme_start_time'] - pd.Timedelta(value='6h')
        end_times = pd.to_datetime(cmes.loc[cmes['sc_insitu']==sc]['mo_end_time']).reset_index()['mo_end_time'] + pd.Timedelta(value='6h')

        mask = res['Spacecraft_ID']==spacecraft_ID(my_spacecraft[i], ID_number=True)

        for j in range(len(start_times)):
            
            start = start_times[j].round("h").tz_localize(None)
            end = end_times[j].round("h").tz_localize(None)

            time_mask = (mask.index < end) & (mask.index > start)


            print(start, end, time_mask.sum())
            res[time_mask] = np.nan


#Times
#res['Time'] = res.index
res['Radial_travel_time'] = (res['Time'] - res['Start_time_of_parcel']).dt.total_seconds()/3600/24
res['Persistence_time'] = (res.index - res['Time']).dt.total_seconds()/3600/24
res['Total_time_difference'] = (res.index - res['Start_time_of_parcel']).dt.total_seconds()/3600/24
res['R_difference'] = (res['R'] - res['Start_R_of_parcel'])
# mask = ((res['Spacecraft_ID'] == spacecraft_ID(spacecraft_to_compare, ID_number=True))
#          | (res['Persistence_time'] < 5))
# omni = res[mask]
# omni = omni[~omni.index.duplicated(keep='first')]
#Lat
res['LAT_difference'] = abs(res['LAT'] - res.index.map(omni['LAT']))


########################################################################################

mask = ((res['Spacecraft_ID'] != spacecraft_ID(spacecraft_to_compare, ID_number=True))
         | (res['Persistence_time'] > 20))
res = res[mask]

mask = (res['Persistence_time'] < 27)
res = res[mask]

########################################################################################


y='Persistence_time'
x='LAT_difference'
#x='R_difference'

res['N_Measured'] = res.index.map(omni['N'])

res = res.reset_index(drop=True)


bins = 10
binx = (res[x].max() - res[x].min()) / bins
binx = (res[x].max() - res[x].min()) / bins
biny = (res[y].max() - res[y].min()) / bins

# Initialize metric matrix
metric = np.full((bins, bins), np.nan)
counts = np.full((bins, bins), np.nan)

# Loop over bins
for i in range(bins):
    for j in range(bins):
        
        x_min = res[x].min() + binx * i
        x_max = res[x].min() + binx * (i + 1)
        y_min = res[y].min() + biny * j
        y_max = res[y].min() + biny * (j + 1)

        mask = (
            (res[x] > x_min) & (res[x] <= x_max) &
            (res[y] > y_min) & (res[y] <= y_max)
        )
        subset = res[mask]

        if len(subset) >= 100:  # Need at least 2 points for regression
            x_data = subset['N'].to_numpy()
            y_data = subset['N_Measured'].to_numpy()
            valid = ~np.isnan(x_data) & ~np.isnan(y_data)
            if np.sum(valid) >= 10:
                slope, intercept, r_value, p_value, std_err = linregress(
                    x_data[valid], y_data[valid])
                mae = abs(x_data[valid] -  y_data[valid]).mean()
                #if i==7 and j==4:mae=np.nan
                rmse = np.sqrt(np.sum((x_data[valid] -  y_data[valid])**2)/np.sum(valid)).mean()
                metric[j, i] = mae
                counts[j, i] = int(round(np.sum(valid/100)))


# Define bin centers for labeling
x_centers = [res[x].min() + (i + 0.5) * binx for i in range(bins)]
y_centers = [res[y].min() + (j + 0.5) * biny for j in range(bins)]


from scipy.ndimage import gaussian_filter

# Apply 2D Gaussian smoothing
smoothed_metric = gaussian_filter(metric, sigma=1)  # adjust sigma to control smoothness

# Create the plot
plt.figure(figsize=(7.5, 6))
ax = sns.heatmap(metric, cmap='viridis', cbar_kws={'label': r'MAE [km s$^{-1}$]'}, 
                 xticklabels=np.round(x_centers, 2),
                 yticklabels=np.round(y_centers, 2),
                 linewidths=0.05, linecolor='gray')

# Set axis labels and title
ax.set_xlabel(x)
ax.set_ylabel(y)

#ax.set_xlabel('ΔR [AU]')
ax.set_ylabel('Δt [d]')
ax.set_ylabel('Δθ [°]')

#ax.set_title('2D Map of Linear Regression R')

# Rotate tick labels for readability
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
ax.invert_yaxis()

# --- Annotate each heatmap square with bin counts ---
for i in range(metric.shape[0]):
    for j in range(metric.shape[1]):
        ax.text(j + 0.3, i + 0.2, f'{counts[i, j]:.0f}',
                ha='center', va='center', color='white', fontsize=12)


plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

# Create LAT and TIME grids
LAT = x_centers
TIME = y_centers
LAT_grid, TIME_grid = np.meshgrid(LAT, TIME)

# Compute MAE for each combination
MAE_LAT = 0.03 * LAT_grid# + 70
MAE_TIME = 0.23 * np.log(TIME_grid + 1)  # +1 to avoid log(0)
#MAE_LAT = TIME_grid *0  # +1 to avoid log(0)

MAE_LAT_TIME = MAE_LAT + MAE_TIME #+ 38.6


x = metric.flatten()
y = MAE_LAT_TIME.flatten()
mask = (~np.isnan(x)) & (x < 5)


slope, intercept, r_value, p_value, std_err = linregress(y[mask], x[mask])

# Compute intercept standard error
x_mean = np.mean(x[mask])
Sxx = np.sum((x[mask] - x_mean)**2)
intercept_stderr = std_err * np.sqrt(np.sum(x[mask]**2)/Sxx)

print(f"r_value: {r_value:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"Intercept stderr: {intercept_stderr:.2f}")


x3d = LAT
y3d = TIME
X, Y = np.meshgrid(x3d, y3d, indexing='ij')
z = metric

Xf = X.ravel()
Yf = Y.ravel()
Zf = z.ravel()

mask = np.isfinite(Xf) & np.isfinite(Yf) & np.isfinite(Zf)
Xf, Yf, Zf = Xf[mask], Yf[mask], Zf[mask]


def model_func(xy, a, b, c):
    x, y = xy
    return a * x + b * np.log(y) + c


try:
    popt, pcov = curve_fit(model_func, (Xf, Yf), Zf)
    a_fit, b_fit, c_fit = popt

    perr = np.sqrt(np.diag(pcov)) 
    a_err, b_err, c_err = perr

except RuntimeError:
    print("Curve fitting failed.")

print(f"a_value: {a_fit:.2f}")
print(f"b_value: {b_fit:.2f}")
print(f"c_value: {c_fit:.2f}")

fitted_func = a_fit*LAT_grid+b_fit*np.log(TIME_grid)+c_fit

x = metric.flatten()
y = fitted_func.flatten()
mask = ~np.isnan(x)

slope, intercept, r_value, p_value, std_err = linregress(y[mask], x[mask])
print(f"r_value: {r_value:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"Intercept stderr: {intercept_stderr:.2f}")



# Plot heatmap
plt.figure(figsize=(7.5, 6))
ax = sns.heatmap(
    MAE_LAT_TIME,
    cmap='viridis',
    cbar_kws={'label': 'MAE [km/s]'},
    xticklabels=np.round(LAT, 1),
    yticklabels=np.round(TIME, 1),
    linewidths=0.05,
    linecolor='gray'
)

# Labels and formatting
ax.set_xlabel('LAT [°]')
ax.set_ylabel('TIME [d]')
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
ax.invert_yaxis()

plt.tight_layout()
plt.show()
```
